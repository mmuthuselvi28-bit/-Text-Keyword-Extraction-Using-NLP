```python
import re

print("===== TEXT KEYWORD EXTRACTOR =====")

text = input("Enter your paragraph:\n")

# Remove punctuation and convert to lowercase
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

# Common words to ignore
stop_words = {
    "the", "is", "a", "an", "and", "of", "to", "in",
    "for", "on", "with", "this", "that", "are", "was",
    "it", "as", "by", "from", "be", "or"
}

# Remove stop words
keywords = [word for word in words if word not in stop_words]

# Count word frequency
frequency = {}

for word in keywords:
    frequency[word] = frequency.get(word, 0) + 1

# Sort by frequency
result = sorted(frequency.items(), key=lambda x: x[1], reverse=True)

print("\n===== KEYWORDS =====")

if result:
    for word, count in result[:10]:
        print(word, ":", count, "time(s)")
else:
    print("No keywords found.")
```
